```python
import sys
sys.path.insert(0, '..')
import os

from src.encoding.cayenne import encode_o3, decode_o3, flip_bit
from src.encoding.crypto import encrypt, decrypt

KEY = os.urandom(16)
DEVADDR, FCNT = 0x26011BDA, 42

trama = encode_o3(43)
print(f"Trama:     {trama.hex(' ')}")

cifrada = encrypt(trama, KEY, DEVADDR, FCNT)
print(f"Cifrada:   {cifrada.hex(' ')}")

atacada = flip_bit(cifrada, 5)          # bit 5 = 32 ppb
recibida = decrypt(atacada, KEY, DEVADDR, FCNT)
print(f"Recibido:  {decode_o3(recibida)} ppb   (esperado 75)")
```
